```python
from langchain_chroma import Chroma
from langchain_ollama import OllamaEmbeddings


def load_chroma_db(model_name='dengcao/Qwen3-Embedding-0.6B:Q8_0',temperature=0,collection_name=None,context_len=2048):
    embeddings = OllamaEmbeddings(
        model=model_name,
        temperature=temperature,
        num_ctx = context_len,
    )
    collection_name = collection_name + '_' + model_name.replace(':','-')

    vector_store = Chroma(
        embedding_function=embeddings,
        collection_name="chat_documents_whatsapp_Qwen3-Embedding-0.6B-Q8_0-latest",#collection_name,
        persist_directory='.chroma_db',
    )
    return vector_store
```
